## Importaciones y configuración

```python
from markitdown import MarkItDown
from google.colab import files
from IPython.display import Markdown, display

md = MarkItDown()

def mostrar_resultado(resultado, titulo='Resultado'):
    """Muestra el texto convertido de forma legible."""
    print(f'--- {titulo} ---')
    print(f'Caracteres: {len(resultado.text_content)}')
    print()
    display(Markdown(resultado.text_content))

print('Listo!')
```

---
## 1. PDF

```python
# Opcion A: subir archivo PDF desde tu computadora
uploaded = files.upload()
pdf_path = list(uploaded.keys())[0]

resultado_pdf = md.convert(pdf_path)
mostrar_resultado(resultado_pdf, 'PDF')
```

```python
# Opcion B: PDF desde URL publica
url_pdf = 'https://www.w3.org/WAI/WCAG21/wcag21.pdf'  # cambia por tu URL

resultado_pdf_url = md.convert(url_pdf)
mostrar_resultado(resultado_pdf_url, 'PDF desde URL')
```

```python
# Guardar resultado PDF en archivo .md
with open('resultado_pdf.md', 'w', encoding='utf-8') as f:
    f.write(resultado_pdf.text_content)

files.download('resultado_pdf.md')
```

```python
# Opcion C: HTML desde string directo
html_string = """
<html>
  <body>
    <h1>Mi titulo</h1>
    <p>Este es un parrafo con <strong>texto en negrita</strong>.</p>
    <ul>
      <li>Item 1</li>
      <li>Item 2</li>
    </ul>
  </body>
</html>
"""

with open('temp.html', 'w') as f:
    f.write(html_string)

resultado_html_string = md.convert('temp.html')
mostrar_resultado(resultado_html_string, 'HTML desde string')
```

```python
# Guardar resultado HTML en archivo .md
with open('resultado_html.md', 'w', encoding='utf-8') as f:
    f.write(resultado_html.text_content)

files.download('resultado_html.md')
```

## 8. JSON

```python
# Opcion A: JSON desde archivo local
uploaded = files.upload()  # sube tu archivo .json
json_path = list(uploaded.keys())[0]

resultado_json = md.convert(json_path)
mostrar_resultado(resultado_json, 'JSON desde archivo')
```

```python
# Opcion B: JSON desde string/diccionario Python
import json

datos = {
    "nombre": "Maria Garcia",
    "edad": 30,
    "ciudad": "Madrid",
    "habilidades": ["Python", "Machine Learning", "SQL"],
    "contacto": {
        "email": "maria@ejemplo.com",
        "telefono": "+34 600 000 000"
    }
}

with open('datos.json', 'w', encoding='utf-8') as f:
    json.dump(datos, f, ensure_ascii=False, indent=2)

resultado_json_dict = md.convert('datos.json')
mostrar_resultado(resultado_json_dict, 'JSON desde diccionario Python')
```

```python
# Guardar resultado JSON en archivo .md
with open('resultado_json.md', 'w', encoding='utf-8') as f:
    f.write(resultado_json_dict.text_content)

files.download('resultado_json.md')
```

## Extra: Convertidor Universal (detecta el tipo automaticamente)

```python
# Sube CUALQUIER archivo y lo convierte automaticamente
import os

print('Sube tu archivo (PDF, DOCX, XLSX, PPTX, JPG, PNG, MP3, HTML, JSON):')
uploaded = files.upload()
archivo = list(uploaded.keys())[0]
extension = os.path.splitext(archivo)[1].lower()

TIPOS_SOPORTADOS = ['.pdf', '.docx', '.doc', '.xlsx', '.xls',
                    '.pptx', '.ppt', '.jpg', '.jpeg', '.png',
                    '.gif', '.webp', '.mp3', '.wav', '.m4a',
                    '.html', '.htm', '.json', '.csv', '.xml']

if extension in TIPOS_SOPORTADOS:
    resultado = md.convert(archivo)
    nombre_salida = os.path.splitext(archivo)[0] + '.md'

    print(f'Archivo: {archivo}')
    print(f'Tipo detectado: {extension}')
    print(f'Caracteres generados: {len(resultado.text_content)}')
    print()
    display(Markdown(resultado.text_content))

    with open(nombre_salida, 'w', encoding='utf-8') as f:
        f.write(resultado.text_content)

    print(f'Descargando: {nombre_salida}')
    files.download(nombre_salida)
else:
    print(f'Tipo "{extension}" no reconocido. Tipos soportados: {TIPOS_SOPORTADOS}')
```
